## Tipos de datos
Para analizar los datos, es importante saber qué tipo de datos estamos tratando.

Podemos dividir los tipos de datos en tres categorías principales:
- Numérico
- Categoría
- Ordinal
  
Los datos **numéricos** son números, y se pueden dividir en dos Categorías numéricas:

- Datos Discretos
    - datos contados que se limitan a enteros. Ejemplo: El número De los coches que pasan.
- Datos Continuos
    - datos medidos que pueden ser cualquier número. Ejemplo: El Precio de un artículo, o el tamaño de un artículo

Los datos **categoricos** son valores que no se pueden medir Uno contra el otro. Ejemplo: un valor de color, o cualquier valor sí/no.

Los datos **ordinales** son como datos categóricos, pero se pueden medir Enfrentarse unos contra otros. Ejemplo: calificaciones escolares donde A es mejor que B y así En

## Mediana, media, modo
En el aprendizaje automático (y en matemáticas) a menudo hay tres valores que Interesándonos:

- Media - El valor medio
-Mediana - El valor de punto medio
- Modo - El valor más común

Ejemplo: Hemos registrado la velocidad de 13 coches:



````python
speed = [99,86,87,88,111,86,103,87,94,78,77,85,86] 
````

### **Media**

El valor medio es el valor medio.

````python
sum = 0
for i in range(len(speed)):
    sum += speed[i]
print("Average speed is:", sum/len(speed))
````

Con **NumPy**

Utilice el `mean()` para encontrar el Velocidad media:

````python
import numpy as np
x = np.mean(speed)
print("Average speed is:", x)
````

### Mediana

El valor medio es el valor en el medio, después de haber ordenado todos los valores.

Con **NumPy**:
Utilice el `median()` para encontrar el Velocidad media:

````python
x = np.median(speed)
print("Median speed is:", x)
````

### Modo
El valor de modo es el valor que aparece más veces.
El módulo **SciPy** tiene un método para esto: 


````python
from scipy import stats

x = stats.mode(speed)
mode = x.mode
print("Mode speed is:", mode)
count = x.count
print("Mode count is:", count)
````

## Desviación Standard
La desviación estándar es un número que describe cuán extendidos están los valores. Una baja desviación estándar significa que la mayoría de los números están cerca del valor medio (promedio), en cambio, una desviación estándar alta significa que los valores se distribuyen en un rango más amplio.

Ejemplo: Esta vez hemos registrado la velocidad de 7 coches:

```speed = [86,87,88,86,87,85,86]```

La desviación estándar es:

```0.9```

Lo que significa que la mayoría de los valores están dentro del rango de 0.9 de la media Valor, que es 86.4.

El módulo **NumPy** tiene un `std()` para calcular la desviación estándar:


````python
speed = [86,87,88,86,87,85,86]
x = np.std(speed)
print("Standard deviation is:", x)
````

## Varianza
La varianza es una medida de dispersión que indica cuánto se separan los valores respecto a la media.
Para calcularla:

1. Calcular la media:

mu = (32 + 111 + 138 + 28 + 59 + 77 + 97) / 7 = 77.4

2. Calcular la diferencia respecto a la media para cada valor:

- 32 - 77.4 = -45.4
- 111 - 77.4 = 33.6
- 138 - 77.4 = 60.6
- 28 - 77.4 = -49.4
- 59 - 77.4 = -18.4
- 77 - 77.4 = -0.4
- 97 - 77.4 = 19.6

3. Elevar al cuadrado cada diferencia:

- (-45.4)^2 = 2061.16
- (33.6)^2 = 1128.96
- (60.6)^2 = 3672.36
- (-49.4)^2 = 2440.36
- (-18.4)^2 = 338.56
- (-0.4)^2 = 0.16
- (19.6)^2 = 384.16

4. La varianza es la media de esas diferencias al cuadrado:

sigma^2 = (2061.16 + 1128.96 + 3672.36 + 2440.36 + 338.56 + 0.16 + 384.16) / 7 = 1432.2

Con NumPy puede calcularse usando `np.var()`

````python
speed = [32,111,138,28,59,77,97]

x = np.var(speed)
print("Variance is:", x)
````

> Como hemos aprendido, la fórmula para encontrar la desviación estándar es la raíz cuadrada de la varianza:

√1432.25 = 37.85 


````python
speed = [32,111,138,28,59,77,97]

x = np.std(speed)

print(x)
````

## Percentiles

Los percentiles indican el valor por debajo del cual se encuentra un porcentaje determinado de los datos.

Ejemplo: Digamos que tenemos una matriz que contiene las edades de cada persona

```ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]```

¿Qué es el percentil 75? La respuesta es 43, lo que significa que el 75% de las personas Son 43 o menos.


Utilice el **NumPy** `percentile()` para encontrar los percentiles:


````python
ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]
x = np.percentile(ages, 75)
print("75th percentile is:", x)
````
